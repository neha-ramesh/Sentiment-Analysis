# **PRE-PROCESSING**

```python
import librosa
import numpy as np
import scipy.signal as sps

SAMPLE_RATE = 16000
DURATION = 3
AUDIO_LENGTH = SAMPLE_RATE * DURATION

def load_and_fix_audio(path, target_sr=SAMPLE_RATE, duration=DURATION):
    # Load mono, resample
    y, sr = librosa.load(path, sr=target_sr, mono=True)

    # Fix length (pad or cut)
    if len(y) > AUDIO_LENGTH:
        y = y[:AUDIO_LENGTH]
    else:
        y = librosa.util.fix_length(y, size=AUDIO_LENGTH)
    return y
```

RMS

```python
def rms_normalize(y):
    rms = np.sqrt(np.mean(y**2))
    if rms > 0:
        return y / rms
    return y
```

Bandpass Filter (100Hz to 8000Hz)

```python
def bandpass_filter(y, sr=SAMPLE_RATE, lowcut=300, highcut=7500):
    sos = sps.butter(10, [lowcut, highcut], btype='band', fs=sr, output='sos')
    return sps.sosfilt(sos, y)
```

Silence Removal using Energy Threshold

```python
def remove_silence(y, sr=SAMPLE_RATE, frame_length=2048, hop_length=512, top_db=30):
    # Non-silent intervals
    intervals = librosa.effects.split(y, top_db=top_db, frame_length=frame_length, hop_length=hop_length)

    if len(intervals) == 0:
        return np.array([])  # All silence

    # Concatenate non-silent parts
    non_silent = np.concatenate([y[start:end] for start, end in intervals])

    # Fix back to 3 seconds
    if len(non_silent) > AUDIO_LENGTH:
        non_silent = non_silent[:AUDIO_LENGTH]
    else:
        non_silent = librosa.util.fix_length(non_silent, size=AUDIO_LENGTH)

    return non_silent
```

Voice Activity Detection

```python
def voice_activity_detect(y, frame_length=2048, hop_length=512, top_db=30):
    # Remove silent frames based on energy threshold
    intervals = librosa.effects.split(y, top_db=top_db, frame_length=frame_length, hop_length=hop_length)
    if len(intervals) > 0:
        return np.concatenate([y[start:end] for start, end in intervals])
    return y  # fallback if silence detection fails
```

Full Preprocessing Wrapper

```python
def full_preprocess(path):
    y = load_and_fix_audio(path)
    y = bandpass_filter(y)
    y = rms_normalize(y)
    #y = remove_silence(y)
    y = voice_activity_detect(y)
    y = librosa.util.fix_length(y, size=AUDIO_LENGTH)
    return y
```

```python
def preprocess_dataset(df, emotion_col="emotion", path_col="path"):
    X = []
    y = []
    failed = 0

    for row in df.itertuples():
        try:
            y_audio = full_preprocess(getattr(row, path_col))
            if len(y_audio) == AUDIO_LENGTH:  # Ensure it's usable
                X.append(y_audio)
                y.append(getattr(row, emotion_col))
            else:
                failed += 1
        except Exception as e:
            failed += 1

    total = len(df)
    success = total - failed
    success_pct = (success / total) * 100
    failed_pct = 100 - success_pct

    print(f"✅ {success}/{total} processed successfully ({success_pct:.2f}%)")
    print(f"❌ {failed} failed ({failed_pct:.2f}%)\n")

    return X, y
```

Pitch shift, Time stretch, Noise injection


```python
import torch
import torchaudio
import torchaudio.transforms as T
import pickle
import random
import os

SAMPLE_RATE = 16000
DURATION = 3

def augment_audio(y, sample_rate=SAMPLE_RATE):
    y_tensor = torch.tensor(y, dtype=torch.float32)

    # Randomly apply pitch shift
    if random.random() < 0.5:
        pitch_shift = T.PitchShift(sample_rate, n_steps=random.choice([-2, -1, 1, 2]))
        y_tensor = pitch_shift(y_tensor)

    # Randomly apply time stretch (only on spectrograms, so do resample instead)
    if random.random() < 0.5:
        stretch_factor = random.uniform(0.85, 1.15)
        y_tensor = torchaudio.functional.resample(y_tensor, sample_rate, int(sample_rate * stretch_factor))
        y_tensor = y_tensor[:SAMPLE_RATE * DURATION]  # Ensure same length
        if y_tensor.shape[0] < SAMPLE_RATE * DURATION:
            pad = SAMPLE_RATE * DURATION - y_tensor.shape[0]
            y_tensor = torch.nn.functional.pad(y_tensor, (0, pad))

    # Randomly add noise
    if random.random() < 0.5:
        noise = 0.005 * torch.randn_like(y_tensor)
        y_tensor += noise

    # Return as NumPy array
    return y_tensor.detach().numpy()
```

# **FEATURE EXTRACTION**

```python
import librosa
import numpy as np
import pywt
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Flatten
from tensorflow.keras.callbacks import EarlyStopping
import matplotlib.pyplot as plt
import seaborn as sns
```

Feature Extraction Functions

```python
# --- Feature Extraction Functions ---
def extract_mfcc(y, sr=16000, n_mfcc=40):
    return librosa.feature.mfcc(y=y, sr=sr, n_mfcc=n_mfcc).T  # (time, 40)

def extract_mfdwc(y, sr=16000, n_mfcc=40, wavelet='db4'):
    mfcc = extract_mfcc(y, sr, n_mfcc)
    coeffs = [pywt.dwt(col, wavelet)[0] for col in mfcc.T]  # DWT on each MFCC coefficient
    mfdwc = np.array(coeffs).T
    return mfdwc.T  # match (time, 40)

def extract_mfwpc(y, sr=16000, n_mfcc=40, wavelet='db4', level=2):
    mfcc = extract_mfcc(y, sr, n_mfcc)
    packets = []
    for col in mfcc.T:
        wp = pywt.WaveletPacket(data=col, wavelet=wavelet, mode='symmetric', maxlevel=level)
        node_values = [node.data for node in wp.get_level(level, order='freq')]
        packets.append(np.concatenate(node_values))
    return np.array(packets).T  # (time, N)
```

```python
import pickle
from tqdm import tqdm

mfcc_list = []
mfdwc_list = []
mfwpc_list = []

print("🔁 Step 1: Extracting individual features (MFCC, MFDWC, MFWPC)...")

for y in tqdm(X_train_aug):
    mfcc = extract_mfcc(y)
    mfdwc = extract_mfdwc(y)
    mfwpc = extract_mfwpc(y)

    # Align by trimming to min length
    min_len = min(mfcc.shape[0], mfdwc.shape[0], mfwpc.shape[0])
    mfcc, mfdwc, mfwpc = mfcc[:min_len], mfdwc[:min_len], mfwpc[:min_len]

    mfcc_list.append(mfcc)
    mfdwc_list.append(mfdwc)
    mfwpc_list.append(mfwpc)

# 🔒 Save individually
with open("/content/drive/MyDrive/mfcc_list.pkl", "wb") as f:
    pickle.dump(mfcc_list, f)

with open("/content/drive/MyDrive/mfdwc_list.pkl", "wb") as f:
    pickle.dump(mfdwc_list, f)

with open("/content/drive/MyDrive/mfwpc_list.pkl", "wb") as f:
    pickle.dump(mfwpc_list, f)

print("✅ Feature extraction complete and saved.")
```
